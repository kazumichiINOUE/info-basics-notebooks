# 情報基礎 対面2 ── シグモイド関数の計算

2ビットの偶数・奇数の問題で，0b11 = (1, 1) の出力 y を，シグモイド関数で計算する．z がスライドで手で出した値と同じになるかを確かめる．

**進め方**

- 上のセルから順に，セルの左の ▶（実行ボタン）を押す．コードは書き換えない
- `NameError` が出たら，1 のセルから順に実行し直す
- ほかのエラーが出たら，まず最後の1行を読む．何が起きたかがそこに書いてある

## 1 初期値と入力

次のセルを，そのまま実行する．値に名前を付けただけなので，何も表示されない．

```python
import math

w1 = 1.764   # 重み w₁ の初期値（乱数）
w2 = 0.4     # 重み w₂ の初期値（乱数）
b = 0        # バイアス b の初期値
x1 = 1       # 入力の1ビット目
x2 = 1       # 入力の2ビット目．(1, 1) は 0b11，10進数の 3
```

## 2 シグモイド関数

次のセルを実行する．シグモイド関数を作るだけで，まだ使っていないので，何も表示されない．

- `def` は，関数を作る書き方．`sigmoid` が名前，`(z)` が入力，`return` の後ろが出力
- `math.exp(-z)` は e の −z 乗（e はネイピア数）．スライドの式 y = 1 / (1 + e⁻ᶻ) を，そのまま書いたもの

```python
def sigmoid(z):
    return 1 / (1 + math.exp(-z))
```

**確かめ**　z = 0 のときの値を出す．スライドでは「z = 0 で ちょうど 0.5」だった．次のセルを実行する．

```python
sigmoid(0)
```

## 3 z と y

次のセルを実行する．z は手で出した値と同じか．y は 0〜1 の数か．

```python
z = x1 * w1 + x2 * w2 + b
y = sigmoid(z)
print("z =", round(z, 3))
print("y =", round(y, 3))
```
